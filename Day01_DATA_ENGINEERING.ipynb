```python
# MINI PROJECT : Student Data Explorer

import pandas as pd

# Student Dataset
data = {
    "name":["Arun","Priya","Kavin","Divya","Rahul"],
    "gender":["Male","Female","Male","Female","Male"],
    "department":["CS","IT","CS","ECE","IT"],
    "math_score":[90,85,78,88,92],
    "programming_score":[95,90,80,85,93],
    "attendance_percentage":[85,92,70,88,96]
}
```

```python
# Create DataFrame
df = pd.DataFrame(data)
```

```python
# Calculate total score
df["total_score"] = df["math_score"] + df["programming_score"]
```

```python
# 1. Project Title
print("="*40)
print("STUDENT PERFORMANCE EXPLORER")
print("="*40)
```

```python
# 2. Dataset Overview
print("\nDataset Overview")

print("Total students:",len(df))
print("Total columns:",len(df.columns))
print("Unique departments:",df["department"].nunique())
```

```python
# 3. Department Wise Analysis
print("\nDepartment Analysis")

print(df["department"].value_counts())
```

```python
# 4. Score Analysis
print("\nScore Analysis")

print("Highest math score:",
      df["math_score"].max())

print("Lowest math score:",
      df["math_score"].min())

print("Average math score:",
      df["math_score"].mean())

print("Average programming score:",
      df["programming_score"].mean())

print("Average attendance:",
      df["attendance_percentage"].mean())
```

```python
# 5. Gender Wise Analysis
print("\nGender Wise Analysis")

print(
df.groupby("gender")["total_score"].mean()
)
```

```python
# 6. Top Performers
print("\nTop 5 Students")

top_students = df.sort_values(
    by="total_score",
    ascending=False
)

print(
top_students[
["name","department","total_score"]
].head(5)
)
```

```python
# 7. Attendance Analysis
print("\nAttendance Analysis")

low_attendance = df[
df["attendance_percentage"] < 75
]

if len(low_attendance)>0:
    print(
    low_attendance[
    ["name",
     "department",
     "attendance_percentage"]
    ]
    )

else:
    print(
    "All students attendance above 75%"
    )
```
